## 2. Core Agentic Workflow Patterns

Let's implement the two foundational agentic patterns: **Routing** and **Evaluator-Optimizer (Reflection Loop)**.


```python
# 1. Routing / Intent Classifier Pattern
from pydantic import BaseModel, Field

class RoutingDecision(BaseModel):
    intent: str = Field(description="Category: 'TECHNICAL_SUPPORT', 'BILLING_INQUIRY', or 'SALES_LEAD'")
    confidence: float = Field(description="Confidence score between 0.0 and 1.0")
    routing_reason: str = Field(description="Justification for the chosen route")

def route_inquiry(user_message: str) -> RoutingDecision:
    prompt = f"Analyze and classify this customer message:\n\"{user_message}\""
    resp = client.models.generate_content(
        model="gemini-2.5-flash",
        contents=prompt,
        config=types.GenerateContentConfig(
            response_mime_type="application/json",
            response_schema=RoutingDecision,
            temperature=0.0
        )
    )
    return RoutingDecision.model_validate_json(resp.text)

test_msg = "We want to upgrade our enterprise cluster to 500 nodes and discuss volume discount pricing for next quarter."
route = route_inquiry(test_msg)
print(f"🚦 Routing Destination: {route.intent} (Confidence: {route.confidence})")
print(f"💡 Rationale: {route.routing_reason}")
```

```python
# 2. Evaluator-Optimizer (Reflection & Self-Correction) Pattern

def generate_and_refine_code(specification: str, max_iterations: int = 2) -> str:
    """Generates code, runs an evaluator LLM critic, and refines if defects exist."""
    # Initial Generator
    current_code = client.models.generate_content(
        model="gemini-2.5-flash",
        contents=f"Write clean, type-annotated Python code fulfilling this spec:\n{specification}",
        config=types.GenerateContentConfig(temperature=0.2)
    ).text
    
    for i in range(max_iterations):
        print(f"🔄 Iteration {i+1}: Evaluating code quality...")
        
        # Critic Prompt
        critic_prompt = f"""
You are a Principal Software Security and Performance Reviewer.
Evaluate this code against the spec:

[SPECIFICATION]:
{specification}

[CODE]:
{current_code}

Identify any edge cases, missing error handlers, or performance bottlenecks.
If code is already production-ready, respond with 'STATUS: PERFECT'. Otherwise, provide bullet points of required fixes.
"""
        critique = client.models.generate_content(
            model="gemini-2.5-flash",
            contents=critic_prompt,
            config=types.GenerateContentConfig(temperature=0.0)
        ).text
        
        if "STATUS: PERFECT" in critique:
            print("✨ Code approved as production-ready!")
            break
            
        print(f"📝 Critic Feedback:\n{critique[:200]}...\n")
        
        # Optimizer Refinement
        refine_prompt = f"""
Refine and rewrite the following Python code by fixing every defect identified in the critique:

[ORIGINAL CODE]:
{current_code}

[CRITIQUE]:
{critique}

Provide the complete updated code.
"""
        current_code = client.models.generate_content(
            model="gemini-2.5-flash",
            contents=refine_prompt,
            config=types.GenerateContentConfig(temperature=0.1)
        ).text
        
    return current_code

spec = "Create a thread-safe in-memory rate limiter in Python that enforces a sliding window log algorithm with max N requests per M seconds."
refined_result = generate_and_refine_code(spec)
print("🚀 Final Refined Implementation:")
print(refined_result[:500] + "...")
```

```python
# Defining Tool Functions for the Agent

def get_stock_price(ticker: str) -> dict:
    """Retrieves real-time stock pricing and market capitalization for a given ticker symbol.
    
    Args:
        ticker: The stock ticker symbol (e.g., 'GOOGL', 'AAPL', 'NVDA')
    """
    # Mock data lookup (in production, connects to AlphaVantage / Bloomberg API)
    market_db = {
        "GOOGL": {"price": 182.50, "currency": "USD", "change_24h": "+2.4%", "market_cap": "2.25T"},
        "NVDA": {"price": 128.90, "currency": "USD", "change_24h": "+4.1%", "market_cap": "3.16T"},
        "AAPL": {"price": 224.20, "currency": "USD", "change_24h": "-0.5%", "market_cap": "3.42T"}
    }
    ticker_clean = ticker.upper().strip()
    if ticker_clean in market_db:
        return {"status": "success", "data": market_db[ticker_clean]}
    return {"status": "error", "message": f"Ticker '{ticker}' not found in exchange feed."}

def calculate_portfolio_value(ticker: str, shares: float) -> dict:
    """Calculates the total dollar valuation of a stock holding.
    
    Args:
        ticker: The stock ticker symbol
        shares: Number of shares owned
    """
    price_info = get_stock_price(ticker)
    if price_info.get("status") == "success":
        total_val = price_info["data"]["price"] * shares
        return {"ticker": ticker, "shares": shares, "current_price": price_info["data"]["price"], "total_value_usd": round(total_val, 2)}
    return {"error": f"Could not price ticker {ticker}"}

# Register tools with Gemini
response = client.models.generate_content(
    model="gemini-2.5-flash",
    contents="What is the current price of GOOGL and how much are 150 shares worth?",
    config=types.GenerateContentConfig(
        tools=[get_stock_price, calculate_portfolio_value],
        temperature=0.0
    )
)

print("🛠️ Automated Function Calling Response:")
print(response.text)
```

## 5. Building the "Ask-the-Web" Perplexity-Style Agent

Let's engineer a full Perplexity-style autonomous agent that:
1. Takes a complex research inquiry.
2. Decomposes the inquiry into multi-hop web search queries.
3. Performs live web searches (via simulated web engine / DuckDuckGo / Wikipedia).
4. Extracts and cleans web page snippets.
5. ReAct loop synthesizes a comprehensive response with inline numerical citations `[1]`, `[2]`.
6. Generates formatted **Source Cards** showing titles, URLs, and snippets.


```python
import re
from bs4 import BeautifulSoup
import urllib.parse

class WebSearchSimulator:
    """Simulated real-time search engine with indexed knowledge & dynamic web scraping."""
    
    def __init__(self):
        self.mock_web_db = {
            "diffusion transformer dit vs autoregressive video generation": [
                {
                    "title": "Scalable Diffusion Models with Transformers (DiT)",
                    "url": "https://arxiv.org/abs/2212.09748",
                    "snippet": "Diffusion Transformers (DiT) replace the traditional U-Net backbone with a Transformer operating on latent patches. DiTs achieve state-of-the-art Fréchet Inception Distance (FID) and exhibit strong compute scaling behavior similar to LLMs."
                },
                {
                    "title": "State of Video Generation 2025: DiT vs Autoregressive Models",
                    "url": "https://deeplearning-review.org/video-gen-2025",
                    "snippet": "Leading video models like Google Veo 2 and Sora utilize Spatio-Temporal Diffusion Transformers (DiT). Meanwhile, next-token autoregressive visual models (e.g. VideoPoet, Chameleon) treat video as discrete token sequences, enabling joint audio-video multimodal conditioning."
                }
            ],
            "inference time scaling test time compute o1 deepseek r1": [
                {
                    "title": "Inference-Time Compute Scaling Laws in Reasoning LLMs",
                    "url": "https://ai-research.org/reasoning-compute-scaling",
                    "snippet": "Inference-time scaling shows that allocating test-time compute via Chain-of-Thought (CoT) search, Monte Carlo Tree Search, and verifier scoring can improve accuracy on complex math and coding benchmarks by over 300% without retraining the base weights."
                },
                {
                    "title": "DeepSeek-R1 Architecture & Training Recipe",
                    "url": "https://deepseek.ai/r1-technical-report",
                    "snippet": "DeepSeek-R1 utilizes Group Relative Policy Optimization (GRPO) to enforce self-verification and exploration during RL, generating thousands of thinking tokens to explore diverse reasoning paths."
                }
            ]
        }

    def search(self, query: str) -> List[Dict[str, str]]:
        """Searches indexed web sources for relevant pages."""
        q_lower = query.lower()
        # Look for partial matches in mock DB
        for key, results in self.mock_web_db.items():
            words = key.split()
            if any(w in q_lower for w in words if len(w) > 3):
                return results
                
        # Default fallback results
        return [
            {
                "title": f"Technical Overview: {query}",
                "url": f"https://tech-index.io/search?q={urllib.parse.quote(query)}",
                "snippet": f"Latest developments regarding {query} highlight advancements in efficiency, architectural scaling, and production deployment across cloud platforms."
            }
        ]

web_engine = WebSearchSimulator()
print("🌐 WebSearchSimulator initialized.")
```

```python
class AskTheWebPerplexityAgent:
    """Production Autonomous 'Ask-the-Web' Perplexity-Style Agent."""
    
    def __init__(self, client: genai.Client):
        self.client = client
        self.web_engine = WebSearchSimulator()
        
    def research(self, user_inquiry: str) -> Dict[str, Any]:
        print(f"🔍 [1. Analyzing User Inquiry]: '{user_inquiry}'")
        
        # Step 1: Decompose Inquiry into 2 specific search queries
        decomp_prompt = f"""
Break down this research question into 2 precise web search engine queries:
Question: "{user_inquiry}"
Output format: Exactly two search queries, one per line.
"""
        queries_text = self.client.models.generate_content(
            model="gemini-2.5-flash",
            contents=decomp_prompt,
            config=types.GenerateContentConfig(temperature=0.0)
        ).text
        
        search_queries = [q.strip("- *1234567890. ") for q in queries_text.strip().split("\n") if q.strip()][:2]
        print(f"📡 [2. Generated Search Queries]: {search_queries}")
        
        # Step 2: Execute Web Searches
        all_sources = []
        source_id = 1
        source_map = {}
        
        for q in search_queries:
            results = self.web_engine.search(q)
            for item in results:
                if item["url"] not in [s["url"] for s in all_sources]:
                    item["citation_id"] = source_id
                    all_sources.append(item)
                    source_map[source_id] = item
                    source_id += 1
                    
        print(f"📚 [3. Retrieved {len(all_sources)} Unique Web Sources]")
        
        # Step 3: Format Context for ReAct Synthesis
        context_blocks = []
        for s in all_sources:
            context_blocks.append(
                f"[{s['citation_id']}] Title: {s['title']}\nURL: {s['url']}\nSnippet: {s['snippet']}"
            )
        context_str = "\n\n".join(context_blocks)
        
        # Step 4: Synthesize Answer with Inline Citations
        synthesis_prompt = f"""
You are a world-class AI Research Intelligence Engine (like Perplexity AI).
Synthesize a comprehensive, authoritative response answering the user inquiry using the retrieved sources.

RULES:
1. Every major factual statement MUST be backed by an inline numerical citation referencing the source, e.g. [1] or [2].
2. Structure your answer with clear bold headings, technical depth, and concise bullet points.
3. Compare trade-offs objectively.
4. Do not invent sources.

RETRIEVED SOURCES:
==================================================
{context_str}
==================================================

USER RESEARCH INQUIRY:
{user_inquiry}
"""
        synthesis_resp = self.client.models.generate_content(
            model="gemini-2.5-flash",
            contents=synthesis_prompt,
            config=types.GenerateContentConfig(temperature=0.1)
        )
        
        return {
            "inquiry": user_inquiry,
            "queries": search_queries,
            "answer": synthesis_resp.text,
            "sources": all_sources
        }

# Instantiate and run
ask_agent = AskTheWebPerplexityAgent(client)
result = ask_agent.research("Compare Diffusion Transformers (DiT) vs Autoregressive models for Video Generation.")

print("
" + "="*80)
print(f"💡 RESEARCH REPORT: {result['inquiry']}")
print("="*80)
print(result['answer'])
print("
" + "-"*40 + " [VERIFIED SOURCE CARDS] " + "-"*40)
for src in result['sources']:
    print(f"[{src['citation_id']}] {src['title']}")
    print(f"    🔗 {src['url']}")
    print(f"    📄 {src['snippet'][:100]}...
")
```

## 7. Step-by-Step Solutions & Architectural Explanations


```python
# ============================================================================
# SOLUTION 3.1: Resilient ReAct Loop with Error Recovery
# ============================================================================
def execute_tool_safely(func, **kwargs):
    """Wrapper that safely executes tools with error reporting."""
    try:
        return {"status": "success", "result": func(**kwargs)}
    except Exception as e:
        return {"status": "error", "error_type": type(e).__name__, "message": str(e)}

# Mock tool with intentional failure condition
def fetch_database_records(table_name: str, limit: int = 10) -> list:
    """Queries an internal SQL database table."""
    valid_tables = ["users", "orders", "analytics"]
    if table_name not in valid_tables:
        raise ValueError(f"Table '{table_name}' does not exist! Valid tables: {valid_tables}")
    return [{"id": i, "record": f"data_{i}"} for i in range(min(limit, 3))]

# Test error recovery prompt
error_test_prompt = """
The user asked for records from the 'transactions' table.
Tool execution returned: {'status': 'error', 'message': "Table 'transactions' does not exist! Valid tables: ['users', 'orders', 'analytics']"}
Adjust your plan and select the best alternative table to fulfill the intent.
"""

recovery_response = client.models.generate_content(
    model="gemini-2.5-flash",
    contents=error_test_prompt,
    config=types.GenerateContentConfig(temperature=0.0)
)
print("🛡️ Agent Error Recovery Plan:")
print(recovery_response.text)
```

```python
# ============================================================================
# SOLUTION 3.2: Parallel Sectioning & Synthesis Workflow
# ============================================================================
import concurrent.futures

def research_subtopic(subtopic: str) -> str:
    """Researches a single topic segment in parallel."""
    prompt = f"Provide a concise, 2-paragraph expert technical summary on: {subtopic}"
    resp = client.models.generate_content(
        model="gemini-2.5-flash",
        contents=prompt,
        config=types.GenerateContentConfig(temperature=0.2)
    )
    return f"### {subtopic}\n{resp.text}"

subtopics = [
    "Zero-Trust Architecture in Google Cloud & Workload Identity Federation",
    "Cloud Armor Adaptive Protection & DDoS Defense",
    "Generative AI Security & SAIF Framework"
]

print("⚡ Launching Parallel Sectioning Tasks...")
with concurrent.futures.ThreadPoolExecutor(max_workers=3) as executor:
    results = list(executor.map(research_subtopic, subtopics))

print("✅ Parallel Execution Complete. Fused Dossier Preview:")
fused_report = "\n\n".join(results)
print(fused_report[:400] + "...")
```

```python
# ============================================================================
# SOLUTION 3.3: Citation Verification & Hallucination Auditor
# ============================================================================
class CitationAudit(BaseModel):
    total_citations_found: int
    invalid_citation_indices: List[int]
    grounded_claims: List[str]
    ungrounded_claims: List[str]
    integrity_score: float = Field(description="Percentage 0.0-1.0 of valid, supported citations")

def verify_citation_integrity(answer: str, sources: List[Dict[str, Any]]) -> CitationAudit:
    # 1. Regex find all [N]
    cited_ids = [int(num) for num in re.findall(r'\[(\d+)\]', answer)]
    valid_source_ids = [s["citation_id"] for s in sources]
    
    invalid = [cid for cid in cited_ids if cid not in valid_source_ids]
    
    # 2. Run LLM Grounding Check
    audit_prompt = f"""
Audit this text and verify if the numerical citations accurately reflect the referenced sources:

[TEXT]:
{answer}

[SOURCES]:
{sources}

Identify grounded vs ungrounded claims.
"""
    resp = client.models.generate_content(
        model="gemini-2.5-flash",
        contents=audit_prompt,
        config=types.GenerateContentConfig(
            response_mime_type="application/json",
            response_schema=CitationAudit,
            temperature=0.0
        )
    )
    return CitationAudit.model_validate_json(resp.text)

audit_result = verify_citation_integrity(result['answer'], result['sources'])
print("🔍 Citation Integrity Audit Report:")
print(f"• Total Citations: {audit_result.total_citations_found}")
print(f"• Integrity Score: {audit_result.integrity_score * 100:.1f}%")
print(f"• Invalid Indices: {audit_result.invalid_citation_indices if audit_result.invalid_citation_indices else 'None'}")
```
